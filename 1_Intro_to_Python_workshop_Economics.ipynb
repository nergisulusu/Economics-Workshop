## Part 1: Macroeconomic Data Precision & Consumer Price Index Math [35 Mins]

### Economic Context: Real vs. Nominal Value Adjustment
In economic analysis and policy formulation, comparing nominal values across time (e.g. GDP, wages, or household income) leads to misleading conclusions due to inflation. Economists express values in **real terms** using the Consumer Price Index (CPI):

$$\text{Real Value} = \text{Nominal Value} \times \left( \frac{\text{CPI}_{\text{base}}}{\text{CPI}_{\text{current}}} \right)$$

Additionally, calculating percentage CPI changes over time allows us to measure annual or multi-period inflation rates:

$$\text{CPI Change (\%)} = \left( \frac{\text{CPI}_{\text{current}} - \text{CPI}_{\text{base}}}{\text{CPI}_{\text{base}}} \right) \times 100$$

### The Floating-Point Representation Trap in Economics
Standard Python floats represent numbers using binary floating-point representations (`0.1 + 0.2` becomes `0.30000000000000004`). When aggregating national accounting totals or multi-year CPI adjustments across millions of households, cumulative rounding errors fail academic rigor and national audit standards. Economists use Python's built-in `decimal` module for exact precision arithmetic.

```python
# Worked Example: Date subtraction and float vs. Decimal CPI adjustment
import datetime
from decimal import Decimal

# 1. Parsing economic survey dates
base_date = datetime.date(2020, 1, 15)
current_date = datetime.date(2026, 6, 15)
days_elapsed = (current_date - base_date).days
print(f"Time elapsed between economic surveys: {days_elapsed} days")

# 2. Nominal wages and CPI figures
nominal_wage = 50_000.0
cpi_base = 100.0
cpi_current = 124.75

# Float calculation
real_wage_float = nominal_wage * (cpi_base / cpi_current)

# Exact Decimal calculation (pass numbers as strings to avoid float approximations)
wage_dec = Decimal("50000.00")
cpi_base_dec = Decimal("100.00")
cpi_curr_dec = Decimal("124.75")
real_wage_dec = wage_dec * (cpi_base_dec / cpi_curr_dec)

print(f"Float Real Wage:   ${real_wage_float:,.10f}")
print(f"Decimal Real Wage: ${real_wage_dec:,.10f}")
print(f"Discrepancy (cents): {(real_wage_float - float(real_wage_dec)) * 100:.4f} cents")
```

## Part 2: Policy Indicator Classification & Economic Data Routing [45 Mins]

### Economic Context: Managing Macro Data Pipelines & Revisions
Economic research institutions maintain extensive databases of indicators (e.g. GDP growth, CPI, Unemployment, Trade Balance, Consumer Sentiment). Databases are updated quarterly as series are revised or newly integrated:
1. Policy analysts perform **set operations** to identify newly added indicators and retired indicators.
2. Analysts construct query requests using **comprehensions**.
3. Data engine routers process queries using **Structural Pattern Matching (`match/case`)**, routing requests to appropriate API endpoints (Annual, Quarterly, Monthly) while enforcing data governance rules.

### Key Concepts: Sets, Comprehensions, and Pattern Matching
- **Sets**: Support set difference (`.difference()`), union, and intersection in $O(1)$ lookup time.
- **Comprehensions**: Provide concise Pythonic syntax for dictionary and list operations.
- **Structural Pattern Matching (`match/case`)**: Unpacks dictionary structures cleanly to replace legacy `if-elif-else` code blocks.

```python
# Worked Example: Indicator set comparison and query routing
q1_indicators = {"GDP_US", "CPI_US", "UNEMP_US", "IND_PROD"}
q2_indicators = {"GDP_US", "CPI_US", "PAYROLLS_US", "RETAIL_SALES"}

added_series = q2_indicators.difference(q1_indicators)
retired_series = q1_indicators.difference(q2_indicators)
print(f"Newly added economic series: {added_series}")
print(f"Retired economic series:    {retired_series}")

# Simple pattern matching demonstration for macro queries
def route_macro_query(query: dict):
    match query:
        case {"indicator": ind, "frequency": "annual", "country": country}:
            return f"Fetching annual data for {ind} ({country})."
        case {"indicator": ind, "frequency": "monthly", "country": country}:
            return f"Fetching monthly data for {ind} ({country})."
        case _:
            return "Rejected: Unknown query format."

print(route_macro_query({"indicator": "GDP_US", "frequency": "annual", "country": "USA"}))
```

### Economic Context: Composite Index Construction & Weight Controls
Economists build composite economic indexes (e.g. Consumer Sentiment Index, Financial Conditions Index, Multi-Factor Economic Health Index) by combining macroeconomic series and survey data using weighted averages:

$$\text{Composite Index} = \sum_{i=1}^{K} w_i \times S_i$$

To prevent biased estimates, indicator sampling weights $w_i$ must sum to exactly $1.0$ (100%). We enforce this domain constraint using a **Custom Domain Exception** (`InvalidEconomicWeightsError`).

```python
# Worked Example: Base class and inheritance for economic indicators

class EconomicIndicator:
    def __init__(self, code: str, category: str):
        self.code = code
        self.category = category
        
    def get_value(self, **kwargs) -> float:
        raise NotImplementedError("Subclasses must implement get_value()")

class SimpleMacroSeries(EconomicIndicator):
    def __init__(self, code: str, value: float):
        super().__init__(code, "Macroeconomic")
        self.value = value
        
    def get_value(self, **kwargs) -> float:
        return self.value

gdp = SimpleMacroSeries("GDP_US", 27.5) # $27.5 Trillion
print(f"{gdp.code} ({gdp.category}): {gdp.get_value()} Trillion USD")
```

## Part 4: Memory-Efficient Economic Data Streaming & ERS Audit [35 Mins]

### Economic Context: High-Frequency Price Feeds & ERS Audit Governance
Macroeconomic policy institutes stream continuous price feeds (e.g. daily energy prices, real-time retail price indexes, agricultural commodity costs). Loading entire historical streams into memory creates system bottlenecks. Economists use **generators** (implementing lazy evaluation via `yield`) to process price ticks item-by-item ($O(1)$ memory complexity).

In line with **Ethics, Responsibility, Sustainability (ERS)** principles, compliance rules require logging price spikes or inflation anomalies to CSV audit files. Using **Context Managers** (`with` statements) and `try/except/finally` blocks guarantees transparent, uncorrupted logging.

```python
# Worked Example: Generators and CSV logging
import csv

# Generator yielding price items
def simple_price_generator(n):
    base_price = 100.0
    for i in range(1, n + 1):
        yield round(base_price + i * 0.5, 2)

gen = simple_price_generator(3)
print(f"Next price from generator: {next(gen)}")
print(f"Next price from generator: {next(gen)}")

# Audit logging with context manager
with open("temp_macro_audit.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Tick", "Price", "Status"])
    writer.writerow([1, 105.50, "INFLATION_ALERT"])

print("Macro audit log created successfully.")
```
